```python
n = 5
for i in range(1, n+1): #row
    for j in range(1, n+1): #star
        print("*", end=" ")
    print()
```

```python
n = 5
for i in range(1, n+1): #row
    for j in range(1, n+1): #star
        if i==1 or i==n or j==1 or j==n:
            print("*", end = " ")
        else:
            print(" ", end = " ")
    print()
```

```python
n = 5
for i in range(1, n+1): #row
    for j in range(n-i): #space
        print(" ", end = " ")
    for k in range(i): #star
        print("*", end = " ")
    print()
    
```
